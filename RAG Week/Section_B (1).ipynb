```python
document = [
    """Artificial Intelligence (AI) is a field of computer science that focuses on
    creating machines and software capable of performing tasks that normally require
    human intelligence. These tasks include learning, reasoning, problem solving,
    perception, and understanding language.

    Machine Learning (ML) is a subset of Artificial Intelligence that allows
    computers to learn patterns from data and make predictions or decisions without
    being explicitly programmed for every task. Machine learning algorithms improve
    their performance by learning from examples.

    Deep Learning is a branch of machine learning that uses artificial neural
    networks with multiple layers. It is especially useful for handling large and
    complex datasets. Deep learning is commonly used in image recognition, speech
    recognition, and natural language processing

    Natural Language Processing (NLP) is a field of AI that enables computers to
    understand, process, and generate human language. NLP is used in applications
    such as chatbots, machine translation, sentiment analysis, text summarization,
    and question answering systems

    Retrieval-Augmented Generation (RAG) is a technique that combines information
    retrieval with a large language model. Instead of depending only on information
    stored in the model, RAG retrieves relevant information from an external
    knowledge source and provides it as context to the language model.

    Vector databases are commonly used in modern RAG systems to store embeddings.
    An embedding represents text as a numerical vector. When a user asks a question,
    the query is converted into an embedding and compared with document embeddings
    to retrieve semantically similar information."""
]

# Function to split the text into smaller chunks
def create_chunks(text, chunk_size=50, overlap=10): #
    words = text.split()
    chunks = []     # Create an empty list to store all the chunks

    start = 0

  # Continue creating chunks until all words are processed
    while start < len(words):
        end = start + chunk_size
        chunk = " ".join(words[start:end])
        chunks.append(chunk)      # Add the created chunk to the chunks list

        start += chunk_size - overlap       # Move to the next chunk while keeping the specified overlap

    return chunks


all_text = " ".join(document)

chunks = create_chunks(all_text)

print("Number of chunks:", len(chunks))

# Go through each chunk and display it
for i, chunk in enumerate(chunks):
    print("\nChunk", i + 1)
    print(chunk)
```

I used a fixed chunk size of 50 words with an overlap of 10 words. This is a  moderate chunk size that keeps enough information in each chunk while avoiding large chunks. The 10-word overlap helps to keep important information

#B2. Dense Retrieval

```python

from sentence_transformers import SentenceTransformer     # Import SentenceTransformer to create text embeddings
from sklearn.metrics.pairwise import cosine_similarity    # Import cosine_similarity to compare similarity between embeddings

model = SentenceTransformer("all-MiniLM-L6-v2")  # Load the pre-trained embedding model

chunk_embeddings = model.encode(chunks)   #Convert text to embedding

def dense_retrieve(query, top_k=3):
    query_embedding = model.encode([query])   # convert user query into embedding

    scores = cosine_similarity(query_embedding,chunk_embeddings)[0]  # find the cosine similarity

    top_indices = scores.argsort()[-top_k:][::-1]

    for index in top_indices:
        print("\nScore:", scores[index])
        print("Chunk:", chunks[index])

```

```python
query1 = "What is ai?"

dense_retrieve(query1)
```
